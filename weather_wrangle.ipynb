# 4. Explode each city’s daily forecast array into one row per (city, country, date). How many total rows?


```python
explode = df.explode("data")
explode
```

```python
data_df = pd.concat([pd.json_normalize(explode["city"]),
                     pd.json_normalize(explode["data"])],
                    axis=1)
```

```python
data_df
```

# 5. For each city, use the mean of temp.day across its forecast horizon and list the top 10 hottest cities.


```python
data_df.groupby(["id", "name", "country"])["temp.day"].mean().sort_values(ascending=False).head(10)
```

# 6. Build a tidy table for San Francisco, California.
Include columns: date, temp_min, temp_max, humidity, weather_description.
Convert temperatures from Kelvin to °C.
Format the dates to be human readable, e.g.: YYYY‑MM‑DD

```python
data_df[(data_df["name"]=="San Francisco")&(data_df["country"]=="US")].id.unique()
```

```python
sf_df = pd.DataFrame()
sf_df[["date", "temp_min", "temp_max", "humidity", "weather"]] = data_df[data_df.id==5391959][["dt", "temp.min","temp.max", "humidity", "weather"]]
```

```python
sf_df.weather.iloc[0]
```

```python
sf_df.weather.apply(len).unique()
```

```python
sf_df["weather_description"] = sf_df.weather.apply(lambda x: x[0]["description"])
```

```python
sf_df["temp_min"]
```

```python
sf_df["temp_min"] = sf_df["temp_min"] - 273.15
sf_df["temp_max"] = sf_df["temp_max"] - 273.15
sf_df["date"] = pd.to_datetime(sf_df["date"], unit="s").dt.strftime("%Y-%m-%d")

# Display the final tidy table
sf_df[["date", "temp_min", "temp_max", "humidity", "weather_description"]]
```
